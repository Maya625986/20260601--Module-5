```python
import pandas as pd

# Load datasets
books = pd.read_csv("data/03_Library Systembook.csv")
customers = pd.read_csv("data/03_Library SystemCustomers (3).csv")

#Preview data
books.head()
customers.head()
```

```python
books.info()
customers.info()

books.isnull().sum()
customers.isnull().sum()
```

```python
books.columns = books.columns.str.strip().str.replace(" ", "_")
customers.columns = customers.columns.str.strip().str.replace(" ", "_")
```

```python
books.columns = books.columns.str.strip().str.replace(" ", "_")
customers.columns = customers.columns.str.strip().str.replace(" ", "_")

print(books.columns)
```

```python
print(books.columns.tolist())
```

```python
books["Book checkout"] = pd.to_datetime(books["Book checkout"], errors= "coerce", dayfirst=True)
books["Book Returned"] = pd.to_datetime(books["Book Returned"], errors="coerce", dayfirst=True)
```

```python
before = len(books)

books = books.dropna(subset=["Customer ID", "Book checkout", "Book Returned"])

after = len(books)

print("Rows removed:", before - after)
```

```python
books = books.drop_duplicates()
customers = customers.drop_duplicates()
```

```python
merged = books.merge(customers, on="Customer ID", how="left")
```

```python
merged["Days_Borrowed"] = (merged["Book Returned"] - merged["Book checkout"]).dt.days
```

```python
merged.to_csv("cleaned_library_data.csv", index=False)
```
